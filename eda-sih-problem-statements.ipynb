```python
#Remove Stopwords
stop_words = set(stopwords.words('english'))

# function to remove stopwords
def remove_stopwords(text):
    no_stopword_text = [w for w in text.split() if not w in stop_words]
    return ' '.join(no_stopword_text)

#Clean Text
def clean_text(text):
    text = text.lower()
    text = re.sub("[^a-zA-Z]"," ",text) 
    text = ' '.join(text.split()) 
    return text

#stemming
stemmer = SnowballStemmer("english")
def stemming(sentence):
    stemSentence = ""
    for word in sentence.split():
        stem = stemmer.stem(word)
        stemSentence += stem
        stemSentence += " "
    stemSentence = stemSentence.strip()
    return stemSentence
df['Problem Statement Title'] = df['Problem Statement Title'].astype(str)
df['Problem Statement Title'] = df['Problem Statement Title'].apply(lambda x: remove_stopwords(x))
df['Problem Statement Title'] = df['Problem Statement Title'].apply(lambda x:clean_text(x))
df['Problem Statement Title'] = df['Problem Statement Title'].apply(stemming)
```

```python
df = df.loc[df['Category'] != 'Software / Hardware']
```

```python
from scipy.stats import chi2_contingency
# Create a contingency table
contingency_table = pd.crosstab(df['Category'], df['Domain Bucket'])

# Calculate the chi-square statistic and p-value
chi2, _, _, _ = chi2_contingency(contingency_table)

# Calculate Cramér's V
n_rows, n_cols = contingency_table.shape
cramers_v = np.sqrt(chi2 / (min(n_rows, n_cols) - 1))

print("Cramér's V:", cramers_v)
```

```python
# Create a contingency table
contingency_table = pd.crosstab(df['Category'], df['Domain Bucket'])

# Create a heatmap using Seaborn
plt.figure(figsize=(10, 6))
sns.heatmap(contingency_table, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Association between Category and Domain Bucket')
plt.xlabel('Domain Bucket')
plt.ylabel('Category')

# Show the plot
plt.tight_layout()
plt.show()
```
